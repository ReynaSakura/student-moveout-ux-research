```python
import pandas as pd

df = pd.read_csv("../data/cleaned_survey.csv")

print(df.shape)
```

```python
df["item_disposition"].value_counts()
```

```python
disposition_counts = (
    df["item_disposition"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
)

disposition_counts
```

```python
disposition_pct = (disposition_counts / len(df) * 100).round(1)

disposition_pct
```

```python
df["first_option"].value_counts()
```

```python
awareness_columns = [
    "awareness_online_marketplaces",
    "awareness_university_groups",
    "awareness_social_media",
    "awareness_friends_students",
    "awareness_charity",
    "awareness_other"
]

awareness_counts = {}

for col in awareness_columns:
    awareness_counts[col] = df[col].notna().sum()

awareness_counts
```

```python
for col in awareness_columns:
    print("\n", col)
    print(df[col].value_counts(dropna=False))
```

```python
awareness_summary = []

for col in awareness_columns:
    counts = df[col].value_counts()
    
    awareness_summary.append({
        "Channel": col,
        "Have used it": counts.get("Have used it", 0),
        "Considered using it, but didn't": counts.get("Considered using it, but didn't", 0),
        "Heard of it, but never considered using it": counts.get("Heard of it, but never considered using it", 0),
        "Never heard of it": counts.get("Never heard of it", 0)
    })

awareness_summary = pd.DataFrame(awareness_summary)

awareness_summary
```
